```python
import os
import json

# 文件路径


# "source_keyword": "西安旅游攻略",
# get date  

content_json_path = r"D:\HKU MDASC 24-26\MediaCrawler\data\xhs\json\search_contents_2025-04-24.json"
comments_json_path = r"D:\HKU MDASC 24-26\MediaCrawler\data\xhs\json\search_comments_2025-04-24.json"
images_folder_path = r"D:\HKU MDASC 24-26\MediaCrawler\data\xhs\images"
output_txt_path = r"D:\HKU MDASC 24-26\MediaCrawler\output_test.txt"

# 加载JSON数据
with open(content_json_path, 'r', encoding='utf-8') as f:
    contents = json.load(f)

with open(comments_json_path, 'r', encoding='utf-8') as f:
    comments = json.load(f)

# 创建一个字典，将评论按note_id组织
comments_by_note_id = {}
for comment in comments:
    note_id = comment["note_id"]
    if note_id not in comments_by_note_id:
        comments_by_note_id[note_id] = []
    comments_by_note_id[note_id].append(comment["content"])

# 生成最终文本内容
with open(output_txt_path, 'w', encoding='utf-8') as output_file:
    for content in contents:
        note_id = content["note_id"]
        title = content["title"]
        desc = content["desc"]
        
        # 写标题
        output_file.write(f"标题：{title}\n")
        
        # 写内容
        output_file.write("内容：\n")
        output_file.write(f"{desc}\n")
        
        # 写评论
        output_file.write("评论：\n")
        if note_id in comments_by_note_id:
            for comment in comments_by_note_id[note_id]:
                output_file.write(f"{comment}\n")
        else:
            output_file.write("暂无评论\n")
        
        # 写OCR识别结果
        output_file.write("OCR识别结果：\n")
        note_folder_path = os.path.join(images_folder_path, note_id)
        if os.path.exists(note_folder_path):
            for txt_file in os.listdir(note_folder_path):
                if txt_file.endswith(".txt"):
                    txt_file_path = os.path.join(note_folder_path, txt_file)
                    with open(txt_file_path, 'r', encoding='utf-8') as tf:
                        ocr_content = tf.read()
                        output_file.write(f"{ocr_content}\n")
        else:
            output_file.write("暂无OCR识别结果\n")
        
        # 添加分隔符
        output_file.write("\n" + "="*50 + "\n\n")

print(f"文本已成功保存到：{output_txt_path}")
```
